## Data Cleaning

1. Remove Duplicated value

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

2. Cleaning Name Columns

```python
df['Name']=df['Name'].str.strip().str.title()
```

3. Fix Negative Billing amount 

```python
df.loc[df['Billing Amount']<0, 'Billing Amount']=df['Billing Amount'].mean()
```

```python
df.describe()
```

```python
df.head()
```

```python
df['Date of Admission']=pd.to_datetime(df['Date of Admission'])
```

```python
df['Discharge Date']=pd.to_datetime(df['Discharge Date'])
```

```python
df.dtypes
```

```python
df['length_of_stay']=(df['Discharge Date']-df['Date of Admission']).dt.days
```

## Data Analysis And Data Visualization

1. Which  Medical Condition affects the most patient across the networks??

```python
condition_count= df['Medical Condition'].value_counts()
condition_count
```

```python
plt.figure(figsize=(8,5))
plt.bar(condition_count.index,condition_count.values,color='#7833FE')
plt.title('Patients by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

2. How is Patient age distributed across network?

```python
average_age=df['Age'].mean()
average_age
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'],bins=20,color='Red',edgecolor='black')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.tight_layout()
plt.show()
```

3. How has average treatment cost change over time ?

```python
average_cost=df['Billing Amount'].mean()
average_cost
```

```python
df['Admission_month'] = df['Date of Admission'].dt.to_period('M')
monthly_avg_cost=df.groupby('Admission_month')['Billing Amount'].mean()
monthly_avg_cost
```

```python
plt.figure(figsize=(10,5))
plt.plot(monthly_avg_cost.index.astype(str), monthly_avg_cost.values,color='#43fc33',marker='o',markersize=5)
plt.xlabel('Month')
plt.ylabel('Average Billing Amount')
plt.title('Monthly Treatment Cost Per Time(Monthly)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

4. Is there a relationship between patient age and treatment cost??

```python
senior_avg_cost=df.loc[df['Age']>=60,'Billing Amount'].mean()
senior_avg_cost
```

```python
sample_df=df.sample(1000,random_state=1)
plt.figure(figsize=(8,5))
plt.scatter(sample_df['Age'],sample_df['Billing Amount'],color='Purple',alpha=0.5,s=15)
plt.title('Age vs Treatment Cost')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.tight_layout()
plt.show()
```

5. Which Insurance Provider Covers The Most Patient?

```python
insurance_count=df['Insurance Provider'].value_counts()
insurance_count
```

```python
plt.figure(figsize=(7,7))
plt.pie(insurance_count.values,labels=insurance_count.index,autopct='%1.1f%%',colors=['Red','Green','Violet','Blue','Yellow'])
plt.title('Patients by Insurance Provider')
plt.tight_layout()
plt.show()
```
